```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.size"] = 14
plt.rcParams["lines.linewidth"] = 2.0
```

```python
csv_path = "/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/results/experiments/vit_classification/loss_landscape_convergence_seed_42.csv"
df = pd.read_csv(csv_path)
df
```

```python
drop_last = 1000

x = df["num_samples"].iloc[:-drop_last]
mean = df["delta"].ewm(alpha=0.02, adjust=False).mean().iloc[:-drop_last]
# std = [10 ** 1] * len(mean)

plt.plot(x, mean)

iter_range = 0.25 * np.arange(-10, 20)

for i in iter_range:
    plt.plot(
        x,
        [10 ** i * x**(-1) for x in x],
        color='gray',
        linestyle='dashed',
        alpha=0.2)
    
plt.text(2e3, 2e-5, r'$k^{{-1}}$', fontsize=12, alpha=0.8)
    
plt.xlim(1e3, 1e6)
plt.ylim(7e-6, 3e-2)

# plt.fill_between(x, mean - std, mean + std, alpha=0.2, label="std")
plt.xscale("log")
plt.title("Loss landscape convergence")
plt.xlabel("# samples")
plt.ylabel("Absolute losses difference")
plt.xscale("log")
plt.yscale("log")
# plt.grid(alpha=0.1)
plt.tight_layout()
plt.savefig("/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/figs/loss_landscape_convergence.pdf", bbox_inches="tight")
plt.show()
```
