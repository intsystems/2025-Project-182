```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["font.size"] = 14
plt.rcParams["lines.linewidth"] = 2.0
```

```python
csv_paths = [
    "/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/results/experiments/vit_classification/assumption_1.csv",
    "/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/results/experiments/vit_classification/assumption_1_seed_42.csv",
    "/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/results/experiments/vit_classification/assumption_1_seed_21.csv",
]

dfs = []
for seed, csv_path in zip([307, 42, 21], csv_paths):
    df_tmp = pd.read_csv(csv_path).rename(columns={"grad_norm": f"grad_norm_seed_{seed}"})
    if len(dfs) > 0:
        df_tmp = df_tmp.drop(columns=["num_samples"])
    dfs.append(df_tmp)
    
df = pd.concat(dfs, axis=1)
df
```

```python
grad_norm = np.array(df[["grad_norm_seed_307", "grad_norm_seed_42", "grad_norm_seed_21"]])
mean = grad_norm.mean(axis=1)
std = grad_norm.std(axis=1)
```

```python
x = df["num_samples"].tolist()

plt.plot(x, mean, label="mean")
plt.fill_between(x, mean - std, mean + std, alpha=0.2, label="std")
plt.xscale("log")
plt.legend()
plt.title("Assumption 1 (zeroing gradients)")
plt.xlabel("# samples")
plt.ylabel("Gradient norm")
plt.grid(alpha=0.1)
plt.tight_layout()
plt.savefig("/home/jovyan/shares/SR008.fs2/nkiselev/sandbox/2025-Project-182/code/figs/assumption_1.pdf", bbox_inches="tight")
plt.show()
```
